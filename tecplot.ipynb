```python
import rasterio
import numpy as np
import tecplot as tp
from tecplot.constant import FileFormat

def convert_tif_to_szplt(tif_path, szplt_path):
    # 1. Read TIF file using rasterio
    with rasterio.open(tif_path) as src:
        image = src.read()  # Read all bands
        transform = src.transform
        crs = src.crs
        # Get dimensions
        bands, height, width = image.shape
        
        # Create X and Y coordinates
        cols, rows = np.meshgrid(np.arange(width), np.arange(height))
        x, y = rasterio.transform.xy(transform, rows, cols)
        x = np.array(x)
        y = np.array(y)

    # 2. Prepare Data for Tecplot
    # Assuming single band raster for simplicity (e.g., elevation/intensity)
    # Shape of image is (bands, height, width), we take first band
    z = image[0] 

    # 3. Create Tecplot Dataset
    dataset = tp.data.create_dataset('TIFData', ['X', 'Y', 'Z'], locations=(width, height))
    zone = dataset.add_ordered_zone('Zone', (width, height))
    
    # 4. Assign data
    zone.values('X')[:] = x.flatten(order='F')
    zone.values('Y')[:] = y.flatten(order='F')
    zone.values('Z')[:] = z.flatten(order='F')

    # 5. Save as SZPLT
    tp.data.save_tecplot_plt(szplt_path, dataset=dataset)
    print(f"Successfully converted {tif_path} to {szplt_path}")

# Example Usage
convert_tif_to_szplt('input_raster.tif', 'output_data.szplt')
```
